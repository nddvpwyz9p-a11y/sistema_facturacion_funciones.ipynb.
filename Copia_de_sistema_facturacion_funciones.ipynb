## 📦 Los datos (ya vienen listos)
Un diccionario para los productos y otro para los clientes.

````python
catalogo = {
    "C-010": {"nombre": "Cuaderno cuadriculado 100 hojas", "precio": 7500,  "iva": 0.19},
    "L-020": {"nombre": "Lápiz negro N.2",                 "precio": 1200,  "iva": 0.19},
    "B-030": {"nombre": "Libro escolar de matemáticas",     "precio": 45000, "iva": 0.00},
}

clientes = {
    "900123": {"nombre": "Colegio San Jose",     "tipo": "colegio"},
    "100987": {"nombre": "Laura Restrepo",       "tipo": "detal"},
    "900456": {"nombre": "Oficina Contadores SAS","tipo": "oficina"},
}

print("Productos:", len(catalogo), "· Clientes:", len(clientes))
````

## 🎁 Cómo se lee una función

```
def  nombre_de_la_funcion(parametros):   ← qué recibe
     """Qué hace"""                     ← documentación
     ...cuerpo...                        ← qué hace
     return resultado                    ← qué devuelve (opcional)
```
Una función que solo **muestra** información no necesita `return`.

## 📖 Misión 1 · `ver_catalogo()`

Convierte el diccionario del catálogo en una función que lo muestre.
No recibe nada, no devuelve nada: solo imprime.

````python
def ver_catalogo():
    """Muestra todos los productos del catálogo."""
    print("CATÁLOGO DE PRODUCTOS")
    # ✏️ TU CÓDIGO AQUÍ: recorre catalogo.items() e imprime código, nombre y precio
# 1. Primero creamos el diccionario global
catalogo = {}

# 2. Luego definimos la función
def ver_catalogo():
    """Muestra todos los productos guardados en el catálogo."""
    if not catalogo:
        print("El catálogo está vacío.")
    else:
        print("\n--- 📚 CATÁLOGO DE LA PAPELERÍA ---")
        for codigo, info in catalogo.items():
            print(f"Código: {codigo} | Nombre: {info['nombre']} | Precio: ${info['precio']}")
        print("-----------------------------------")

# 3. Y finalmente la llamamos
ver_catalogo()
````

````python
def agregar_producto(codigo, nombre, precio):
    """Agrega un producto nuevo al catálogo."""
    # ✏️ TU CÓDIGO AQUÍ: guarda el producto y avisa con un print

def agregar_producto(codigo, nombre, precio):
    """Recibe código, nombre y precio, y guarda el producto en el catálogo."""
    catalogo[codigo] = {
        "nombre": nombre,
        "precio": precio
    }
    print(f"✅ Producto '{nombre}' agregado exitosamente.")
agregar_producto("T-040", "Memoria USB 32 GB", 38000)
ver_catalogo()
````

## 👥 Misión 3 · `ver_clientes()`

La misma idea de la misión 1, pero con el diccionario de clientes.

````python
def ver_clientes():
    """Muestra todos los clientes registrados."""
    # ✏️ TU CÓDIGO AQUÍ
# 1. Inicializamos el diccionario global de clientes primero
clientes = {}

# 2. Luego definimos la función de la Misión 3
def ver_clientes():
    """Muestra todos los clientes guardados en el diccionario de clientes."""
    if not clientes:
        print("No hay clientes registrados.")
    else:
        print("\n--- 👥 LISTA DE CLIENTES ---")
        for id_cliente, info in clientes.items():
            print(f"ID: {id_cliente} | Nombre: {info['nombre']} | Teléfono: {info['telefono']}")
        print("----------------------------")

# 3. Y por último la llamamos
ver_clientes()
ver_clientes()
````

## 🗑️ Misión 4 · `eliminar_cliente()`

Recibe el documento y borra ese cliente con `del clientes[documento]`.
Antes de borrar, revisa con un `if` que exista, para que el programa no se caiga.

````python
def eliminar_cliente(documento):
    """Elimina un cliente si existe."""
    # ✏️ TU CÓDIGO AQUÍ: if documento in clientes: ... else: avisa
def eliminar_cliente(documento):
    """Recibe el documento y borra ese cliente si existe en el diccionario."""
    if documento in clientes:
        del clientes[documento]
        print(f"✅ Cliente con documento {documento} eliminado exitosamente.")
    else:
        print(f"❌ Error: El cliente con documento {documento} no existe en el registro.")

eliminar_cliente("900456")
ver_clientes()
````

## 🔍 Misión 5 · `consultar()`

Recibe un dato y lo busca primero en productos y luego en clientes.
Si no está en ninguno, avisa que no lo encontró.

````python
def consultar(dato):
    """Busca el dato en productos o en clientes."""
    # ✏️ TU CÓDIGO AQUÍ: if dato in catalogo: ... elif dato in clientes: ... else: ...
def consultar(dato):
    """Busca un dato primero en el catálogo de productos y luego en clientes."""
    if dato in catalogo:
        info = catalogo[dato]
        print(f"📦 Producto encontrado - Código: {dato} | Nombre: {info['nombre']} | Precio: ${info['precio']}")
    elif dato in clientes:
        info = clientes[dato]
        print(f"👤 Cliente encontrado - Documento: {dato} | Nombre: {info['nombre']} | Teléfono: {info['telefono']}")
    else:
        print(f"❌ El dato '{dato}' no se encuentra registrado ni en productos ni en clientes.")

consultar("C-010")
consultar("900123")
consultar("XXX")
````

## 🖥️ Misión 6 · `menu()` — la integradora

Aquí se juntan todas tus funciones. El menú ya no hace el trabajo: solo **dibuja las opciones**, pregunta y **llama** a la función correspondiente.

Así se verá en pantalla:
```
================================
   EL LÁPIZ FELIZ · CAJA
================================
   1. Ver catálogo
   2. Agregar producto
   3. Ver clientes
   4. Consultar
   5. Salir
```


````python
def menu():
    """Menú de la caja: solo coordina, no calcula."""
    opcion = ""
    while opcion != "5":
        print("=" * 32)
        print("   EL LÁPIZ FELIZ · CAJA")
        print("=" * 32)
        print("   1. Ver catálogo")
        print("   2. Agregar producto")
        print("   3. Ver clientes")
        print("   4. Consultar")
        print("   5. Salir")
        opcion = input("   Opción: ")

        # ✏️ TU CÓDIGO AQUÍ: un if/elif por cada opción, llamando a tus funciones

def menu():
    """Función integradora que muestra el menú principal y gestiona las opciones del usuario."""
    while True:
        print("\n================================")
        print("   EL LÁPIZ FELIZ · CAJA")
        print("================================")
        print("   1. Ver catálogo")
        print("   2. Agregar producto")
        print("   3. Ver clientes")
        print("   4. Consultar")
        print("   5. Salir")
        print("================================")

        opcion = input("Elige una opción (1-5): ")

        if opcion == "1":
            ver_catalogo()

        elif opcion == "2":
            codigo = input("Ingresa el código del producto: ")
            nombre = input("Ingresa el nombre del producto: ")
            precio = float(input("Ingresa el precio del producto: "))
            agregar_producto(codigo, nombre, precio)

        elif opcion == "3":
            ver_clientes()

        elif opcion == "4":
            dato = input("Ingresa el código de producto o documento de cliente a consultar: ")
            consultar(dato)

        elif opcion == "5":
            print("\n¡Gracias por usar el sistema de la Papelería El Lápiz Feliz! 👋")
            break

        else:
            print("❌ Opción inválida. Por favor, elige un número entre 1 y 5.")
menu()
````
